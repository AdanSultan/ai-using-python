```python
# Mini Project: Number Guessing Game

import random

random_number = random.randint(1, 100)
attempts_left = 10

while attempts_left > 0:
    print(f"\nAttempts left: {attempts_left}")

    try:
        guess = int(input("Guess the number between 1 and 100: "))
    except ValueError:
        print("Please enter a whole number.")
        continue

    if not 1 <= guess <= 100:
        print("Please enter a number from 1 to 100.")
        continue

    if guess == random_number:
        print("Congratulations! You guessed the correct number.")
        break

    difference = abs(guess - random_number)

    if difference <= 2:
        if guess < random_number:
            print("Very close! Try a slightly higher number.")
        else:
            print("Very close! Try a slightly lower number.")
    elif guess < random_number:
        print("Too low! Try a higher number.")
    else:
        print("Too high! Try a lower number.")

    attempts_left -= 1
else:
    print("Sorry, you've run out of attempts.")
    print("The correct number was:", random_number)
```

```python
import random


def math_quiz():
    score = 0
    total_questions = 5

    print("Welcome to the Math Quiz!")
    print("Answer the following questions.\n")

    for question_number in range(1, total_questions + 1):
        number1 = random.randint(1, 20)
        number2 = random.randint(1, 20)
        operator = random.choice(["+", "-", "*"])

        if operator == "+":
            correct_answer = number1 + number2
        elif operator == "-":
            correct_answer = number1 - number2
        else:
            correct_answer = number1 * number2

        while True:
            user_answer = input(
                f"Question {question_number}: {number1} {operator} {number2} = "
            )

            try:
                user_answer = int(user_answer)
                break
            except ValueError:
                print("Please enter a valid number.")

        if user_answer == correct_answer:
            print("Correct!\n")
            score += 1
        else:
            print(f"Incorrect. The correct answer was {correct_answer}.\n")

    print("Quiz complete!")
    print(f"Your score: {score}/{total_questions}")

    if score == total_questions:
        print("Excellent work!")
    elif score >= 3:
        print("Good job!")
    else:
        print("Keep practicing!")


math_quiz()
```

```python
import random


def get_number(prompt, minimum, maximum):
    """Get a valid number from the user."""
    while True:
        try:
            number = int(input(prompt))

            if minimum <= number <= maximum:
                return number

            print(f"Please enter a number from {minimum} to {maximum}.")
        except ValueError:
            print("Please enter a valid whole number.")


def create_question(difficulty, operations):
    """Create a random math question and return its answer."""
    if difficulty == "easy":
        maximum = 10
    elif difficulty == "medium":
        maximum = 50
    else:
        maximum = 100

    operator = random.choice(operations)
    first = random.randint(1, maximum)
    second = random.randint(1, maximum)

    if operator == "+":
        answer = first + second

    elif operator == "-":
        # Prevent negative answers
        if second > first:
            first, second = second, first

        answer = first - second

    elif operator == "*":
        answer = first * second

    else:
        # Create a division question with a whole-number answer
        answer = first
        first = first * second

    return f"{first} {operator} {second}", answer


def math_quiz():
    print("\n=== Math Quiz Game ===")

    print("\nChoose a difficulty level:")
    print("1. Easy")
    print("2. Medium")
    print("3. Hard")

    difficulty_choice = get_number("Your choice: ", 1, 3)
    difficulty = ["easy", "medium", "hard"][difficulty_choice - 1]

    print("\nAvailable operations: +, -, *, /")
    operation_text = input(
        "Enter operations separated by spaces, or press Enter for all: "
    ).strip()

    if operation_text:
        operations = operation_text.split()
        operations = [
            operation
            for operation in operations
            if operation in ["+", "-", "*", "/"]
        ]
    else:
        operations = ["+", "-", "*", "/"]

    if not operations:
        print("No valid operations selected. All operations will be used.")
        operations = ["+", "-", "*", "/"]

    total_questions = get_number(
        "\nHow many questions would you like? (1-20): ",
        1,
        20,
    )

    score = 0
    streak = 0
    best_streak = 0
    lives = 3
    questions_answered = 0

    print("\nType 'q' at any time to quit.")
    print("You have 3 lives.\n")

    for question_number in range(1, total_questions + 1):
        question, correct_answer = create_question(
            difficulty,
            operations,
        )

        user_answer = input(
            f"Question {question_number}: {question} = "
        ).strip().lower()

        if user_answer == "q":
            print("\nYou ended the quiz early.")
            break

        questions_answered += 1

        try:
            user_answer = int(user_answer)

            if user_answer == correct_answer:
                streak += 1
                best_streak = max(best_streak, streak)

                # Extra points for consecutive correct answers
                points = 10 + (streak - 1) * 2
                score += points

                print(f"Correct! You earned {points} points.")
                print(f"Current streak: {streak}")

            else:
                lives -= 1
                streak = 0

                print(f"Incorrect. The correct answer was {correct_answer}.")

        except ValueError:
            lives -= 1
            streak = 0

            print("That was not a valid number.")
            print(f"The correct answer was {correct_answer}.")

        print(f"Score: {score} | Lives left: {lives}\n")

        if lives == 0:
            print("Game over! You ran out of lives.")
            break

    print("=== Quiz Results ===")
    print(f"Final score: {score}")
    print(f"Questions answered: {questions_answered}")
    print(f"Best streak: {best_streak}")

    if questions_answered > 0:
        percentage = (score / (questions_answered * 10)) * 100
        print(f"Performance: {percentage:.1f}%")

    if score >= questions_answered * 10 and questions_answered > 0:
        print("Excellent work!")
    elif score > 0:
        print("Good job! Keep practicing to improve.")
    else:
        print("Keep practicing. You will get better!")


while True:
    math_quiz()

    play_again = input(
        "\nWould you like to play again? (y/n): "
    ).strip().lower()

    if play_again != "y":
        print("Thanks for playing!")
        break
```
