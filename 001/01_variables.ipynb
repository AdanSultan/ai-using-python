# **Variables and types**

**Variables**

Variables are named containers used to store data values, while data types define the kind of value a variable holds and how the computer can interact with it. Python is dynamically typed, meaning you do not need to explicitly declare a data type when creating a variable, Python detects it automatically based on the value assigned.

1. Numbers 
2. String
3. Float
4. Boolean
5. Tuple
6. List
7. Dictionary
8. Set
   

# *1. Integer*

```python
age = 23 
print(f"My age is {age} years old")
```

# *2. Floating Point Numbers*

```python
temperature = 28.6
print(type(temperature))
print(f"The temperature is {temperature:.1f} degrees Celsius")
```

# *3. String*

```python
name = "Adan Sultan"
print(f"My name is {name} and my age is {age} years")
```

# *4. Boolean*

```python
is_student = True
print(f"Am I a student: {is_student}")
```

# *5. List*

```python
numbers = [1, 2, 3, 4, 5]

names = ["Alice", "Bob", "Charlie"]
print(type(numbers))
```

# *6. Tuples*

Tuples are immutable 

```python
prices = (9.99, 19.99, 29.99)
print(type(prices))
```

#  *7. Disctionary*

```python
person = {"name": "Alice", "age": 30, "city": "New York"}
print(type(person))
```

# *8. Set*

```python
set_of_numbers = {1, 2, 3, 4, 5}
print(type(set_of_numbers))
```

# *9. None Type*

```python
result = None
print(f"The result is: {result}")
```

# *10. Complex Numbers*

```python
complex_number = 3 + 4j
print(f"The complex number is: {complex_number}")
print(f"The real part is: {complex_number.real} and the imaginary part is: {complex_number.imag}")
```

# *Mutable & Immutable Data Types:*
- Mutable data types can be changed after they are created. Examples include lists, dictionaries, and sets.
- Immutable data types cannot be changed after they are created. Examples include integers, floats, strings, and tuples.
