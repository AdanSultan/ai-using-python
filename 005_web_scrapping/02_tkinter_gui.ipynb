```python
import tkinter as tk

def show_message():
    name = name_entry.get().strip()
    message_label.config(text=f"Hello, {name or 'there'}!")

window = tk.Tk()
window.title("Simple Tkinter GUI")
window.geometry("350x220")

title_label = tk.Label(window, text="Welcome", font=("Arial", 16, "bold"))
title_label.pack(pady=15)

name_entry = tk.Entry(window, width=30)
name_entry.pack(pady=5)
name_entry.insert(0, "Enter your name")

show_button = tk.Button(window, text="Show Message", command=show_message)
show_button.pack(pady=10)

message_label = tk.Label(window, text="")
message_label.pack(pady=5)

window.mainloop()
```
